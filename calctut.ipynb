```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

import matplotlib.pyplot as plt


ar = np.random.randn(10000)
plt.hist(ar, bins = 15)
plt.show()
```

```python
# Generate 100 random values between 0 and 2π
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sin, cos, and tan of the random values
y_sin = np.sin(x)
y_cos = np.cos(x)
y_tan = y_sin/y_cos

# Plot sin, cos, and tan graphs meowwww
plt.figure(figsize=(20,10))

# Plot sine wave
plt.plot(x, y_tan, label='sin(x)', color='blue')
plt.title('Sine Wave')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
```

```python
# Plot cosine wave
plt.figure(figsize=(20,10))

plt.plot(x, y_cos, label='cos(x)', color='green')
plt.title('Cosine Wave')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.grid(True)
```

```python
# Generate values for x between 0.1 and 10 (logarithm can't be zero or negative)
x = np.linspace(0.1, 10, 100)

# Calculate the log and exp of x
y_log = np.log(x)
y_exp = np.exp(x)

# Plot Logarithmic and Exponential graphs
plt.figure(figsize=(10, 6))

# Plot logarithmic curve
plt.subplot(2, 1, 1)
plt.plot(x, y_log, label='log(x)', color='purple')
plt.title('Logarithmic Function')
plt.xlabel('x')
plt.ylabel('log(x)')
plt.grid(True)

# Plot exponential curve
plt.subplot(2, 1, 2)
plt.plot(x, y_exp, label='exp(x)', color='orange')
plt.title('Exponential Function')
plt.xlabel('x')
plt.ylabel('exp(x)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Generate x and y values (equivalent to seq(-3, 3, length=30) in R)
x = np.linspace(-3, 3, 30)
y = x  # y = x

dat = pd.DataFrame({'x': x, 'y': y})

# Plot the line graph
plt.plot(dat['x'], dat['y'], label='y=x')

# Add title
plt.title('y=x')

# Add labels
plt.xlabel('x')
plt.ylabel('y')

# Display the plot
plt.grid(True)
plt.show()
```

```python
# Generate x values and calculate y = -x
x = np.linspace(-3, 3, 30)
y = -x  # y = -x

# Create a pandas DataFrame to store the data
dat = pd.DataFrame({'x': x, 'y': y})

# Plot the line graph with dashed lines (linestyle='--' in Matplotlib)
plt.plot(dat['x'], dat['y'], linestyle='--', label='y=-x')

# Add title
plt.title('y=-x')

# Add labels
plt.xlabel('x')
plt.ylabel('y')

# Display the plot
plt.grid(True)
plt.show()
```

```python
# Generate x values and calculate y = x^2 + 1
x = np.linspace(-3, 3, 30)
y = x**5 + 1  # y = x^2 + 1

# Create a pandas DataFrame to store the data
dat = pd.DataFrame({'x': x, 'y': y})

# Plot the line graph
plt.plot(dat['x'], dat['y'], label='y=x^2 + 1')

# Add title
plt.title('y=x^2 + 1')

# Add labels
plt.xlabel('x')
plt.ylabel('y')

# Display the plot
plt.grid(True)
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# Define x and y values
x = np.linspace(-3, 3, 30)
y = np.linspace(-3, 3, 30)

# Create a meshgrid for X and Y values
X, Y = np.meshgrid(x, y)

# Define the function f(x, y) = 10 - 4x - 5y
Z = 10 - 4*X - 5*Y

# Create the 3D surface plot
fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y)])

# Update layout for better presentation
fig.update_layout(title='3D Surface Plot of f(x, y) = 10 - 4x - 5y',
                  scene = dict(
                      xaxis_title='X-axis',
                      yaxis_title='Y-axis',
                      zaxis_title='Z-axis'),
                  width=800, height=600)

# Display the interactive plot
fig.show()
```

```python
x = np.linspace(-np.pi, np.pi, 100)
y = np.linspace(-np.pi, np.pi, 100)

# Create a meshgrid for X and Y values
X, Y = np.meshgrid(x, y)

# Compute z directly using the function cos(x * y)
Z = np.sin(X * Y)

# Create the 3D surface plot
fig = go.Figure(data=[go.Surface(z=Z, x=x, y=y)])

# Update layout for better presentation
fig.update_layout(title='Surface Plot of cos(x*y)',
                  scene=dict(
                      xaxis_title='X-axis',
                      yaxis_title='Y-axis',
                      zaxis_title='Z-axis'),
                  width=800, height=600)

# Display the interactive plot
fig.show()
```

```python
# Create sequences for x and y
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

# Create a grid of x and y values
X, Y = np.meshgrid(x, y)

# Define the function f(x, y) = log(x^2 + y^2)
Z = np.log(X**2 + Y**2)

# Plot the contour plot using Matplotlib with simple lines
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, colors='black')

# Add labels
plt.title('Contour Plot of f(x, y) = log(x^2 + y^2)')
plt.xlabel('x')
plt.ylabel('y')

# Show the plot
plt.show()
```

```python
# Create sequences for x and y
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

# Create a grid of x and y values
X, Y = np.meshgrid(x, y)

# Define the function f(x, y) = abs(x * y)
Z = np.abs(X * Y)

# Plot the contour plot using Matplotlib
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, colors='black')

# Add labels
plt.title('Contour Plot of f(x, y) = |x * y|')
plt.xlabel('x')
plt.ylabel('y')

# Show the plot
plt.show()
```

```python
# Create sequences for x and y
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

# Create a grid of x and y values
X, Y = np.meshgrid(x, y)

# Define the function f(x, y) = sqrt(36 - 9*x^2 - 4*y^2)
Z = np.sqrt(np.maximum(36 - 9*X**2 - 4*Y**2, 0))  # Use np.maximum to handle values where sqrt argument is negative

# Plot the contour plot using Matplotlib
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, colors='black')

# Add labels
plt.title('Contour Plot of f(x, y) = sqrt(36 - 9*x^2 - 4*y^2)')
plt.xlabel('x')
plt.ylabel('y')

# Show the plot
plt.show()
```
